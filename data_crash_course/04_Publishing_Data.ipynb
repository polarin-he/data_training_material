#### How and Where to Publish Your Data
To integrate your data into **POLARIN**, follow these steps:

## Open Science principles
   - Ensure your data is **FAIR (Findable, Accessible, Interoperable, Reusable)**.
   - Provide **rich metadata** (e.g., title, abstract, keywords, temporal/spatial coverage).
   - Assign a **persistent identifier (PID)** like a DOI.

## Data Licensing
   - Choose an **open license** (e.g., CC-BY 4.0) to maximize reusability.
   - Clearly state the license in your metadata.

## What data repository should I publish my data to?
   Publish your data in a **trusted repository**. Prioritize data centers **actively integrated into POLARIN**:
   - **Biodiversity/Ecological Data (Arctic)**:
     - [ABDS (Arctic Biodiversity Data Service)](https://abds.is/) (CAFF)
   - **Geophysical Data (Arctic)**:
     - [IADC (Italian Arctic Data Center)](https://iadc.cnr.it) (CNR)
   - **Geophysical Data (Antarctic)**:
     - [NADC (National Antarctic Data Center)](https://iandc.pnra.aq) (CNR)
   - **Seismological Data (both polar regions)**:
     - [POSEIDON (GFZ)](http://geofon.gfz-potsdam.de/)
   - **Geophysical and biodiversity/ecological data (Svalbard)**:
     - [SIOS SDMS (Svalbard Integrated Arctic Earth Observing System)](https://sios-svalbard.org/metsis/search)
   - **Geophysical and biodiversity/ecological data (Arctic)**:
     - [Arctic Data Center (MET Norway)](https://adc.met.no/)

   **⚠️ Important**: Always **discuss publication with the data center** to ensure compliance with their requirements.
